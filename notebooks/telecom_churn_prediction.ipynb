# Part B – Comparison of Three AI Techniques

## Technique 1: Logistic Regression

**Overview:** Logistic Regression is a linear classification algorithm that models churn probability using the sigmoid function applied to weighted feature inputs. It estimates the likelihood of customer discontinuation based on demographic and behavioral characteristics, learning optimal feature weights through maximum likelihood estimation via gradient descent.

**Application:** For telecom churn prediction, it identifies linear relationships between customer attributes (tenure, contract type, monthly charges) and churn risk. Each coefficient directly indicates whether a feature increases or decreases churn probability, providing actionable business insights.

**Requirements:** Requires numeric inputs (one-hot encoding for categorical variables) and benefits from feature scaling. Assumes linear relationship between predictors and log-odds of outcome. Training complexity: O(n × m × iterations).

**Output:** Produces probability scores (0-1) and binary predictions. Feature coefficients are directly interpretable for business stakeholders.

**Strengths:** Exceptional interpretability, computational efficiency (fast training/inference), regulatory compliance (GDPR explainability), probabilistic outputs for risk stratification.

**Limitations:** Assumes linear decision boundaries, cannot capture complex non-linear relationships or feature interactions without manual engineering, sensitive to multicollinearity and outliers.

## Technique 2: Decision Tree

**Overview:** Decision Trees create hierarchical structures with branching rules that partition data based on feature values. They recursively split data by maximizing information gain at each node, learning interpretable rules like "IF month-to-month contract AND tenure < 12 months THEN predict churn."

**Application:** For telecom retention, trees automatically discover interaction effects and non-linear patterns without manual feature engineering. They identify nuanced relationships, such as high charges predicting churn only for month-to-month contracts, providing natural customer segmentation.

**Requirements:** Handles mixed data types (categorical and numeric) natively without encoding or scaling. Robust to outliers and missing values. Training complexity: O(n × m × log n).

**Output:** Produces class labels and probability estimates. Feature importance derived from split frequency and depth. Entire decision pathway visualizable for stakeholder communication.

**Strengths:** Captures non-linear relationships and interactions automatically, minimal preprocessing required, highly interpretable visual representation, handles heterogeneous data well.

**Limitations:** Prone to overfitting without pruning/depth limits, high variance (unstable to data changes), biased toward multi-category features, struggles with class imbalance.

## Technique 3: Neural Network (Multi-Layer Perceptron)

**Overview:** Neural Networks consist of interconnected layers of artificial neurons, each applying weighted sums followed by non-linear activation functions (ReLU, sigmoid, tanh). Architecture includes input layer, hidden layers learning intermediate representations, and output layer. Learning occurs via backpropagation, propagating errors backward to update weights through gradient descent.

**Application:** For telecom churn, networks automatically discover hierarchical patterns—early layers detect simple features (high charges), deeper layers identify complex combinations (fiber + month-to-month + e-payment + moderate tenure). Eliminates manual feature engineering.

**Requirements:** All features must be numeric (one-hot encoding required). Critical feature scaling (standardization/normalization). Benefits from large datasets (thousands to millions of samples). Training complexity: O(n × m × h × L × epochs).

**Output:** Probability scores via sigmoid/softmax activation. Internal representations not interpretable ("black box"). Requires SHAP/LIME for post-hoc explainability.

**Strengths:** Learns extremely complex non-linear relationships, state-of-the-art performance with sufficient data, flexible architecture, automatic feature interaction discovery.

**Limitations:** Limited interpretability (GDPR compliance issues), requires large datasets, computationally expensive training, extensive hyperparameter tuning, prone to overfitting without regularization (dropout, L2).

````python
# =============================================================================
# PREPROCESSING FUNCTIONS
# =============================================================================

print("\n" + "=" * 70)
print("PREPROCESSING FUNCTIONS")
print("=" * 70)

#Standard Scaler: z = (x - mean) / std
class Scaler:
    def __init__(self):
        self.mean = None
        self.std = None
        self.mean_ = None   
        self.scale_ = None   
        self.n_samples_seen_ = None
    
    def fit(self, X):
        X = np.asarray(X, dtype=float)
        self.mean = X.mean(axis=0)
        self.std = X.std(axis=0)
        self.std[self.std == 0] = 1.0
        self.mean_ = self.mean
        self.scale_ = self.std
        self.n_samples_seen_ = X.shape[0]
        return self
    
    def transform(self, X):
        X = np.asarray(X, dtype=float)
        return (X - self.mean) / self.std
    
    def fit_transform(self, X):
        self.fit(X)
        return self.transform(X)


class CategoricalEncoder:
    def __init__(self):
        self.categories = {}
        self.feature_names = []
    
    def fit(self, X):
        if not isinstance(X, pd.DataFrame):
            raise TypeError("Input must be a DataFrame")
        
        self.categories.clear()
        self.feature_names.clear()
        
        for col in X.columns:
            values = sorted(X[col].astype(str).unique())
            self.categories[col] = values
            for v in values:
                self.feature_names.append(f"{col}_{v}")
        
        return self
    
    def transform(self, X):
        X = X.copy()
        encoded_cols = []
        
        for col, values in self.categories.items():
            col_data = X[col].astype(str).values
            out = np.zeros((len(col_data), len(values)))
            
            for i, v in enumerate(values):
                out[:, i] = (col_data == v).astype(float)
            
            encoded_cols.append(out)
        
        return np.hstack(encoded_cols)
    
    def fit_transform(self, X):
        self.fit(X)
        return self.transform(X)


def split_data(X, y, test_size=0.2, seed=None, stratify=None):
    if seed is not None:
        np.random.seed(seed)
    
    n = len(X)
    
    if stratify is None:
        indices = np.arange(n)
        np.random.shuffle(indices)
        split_point = int(n * test_size)
        test_idx = indices[:split_point]
        train_idx = indices[split_point:]
    else:
        train_idx = []
        test_idx = []
        
        unique_classes = np.unique(stratify)
        
        for class_label in unique_classes:
            class_indices = np.where(stratify == class_label)[0]
            np.random.shuffle(class_indices)
            n_class = len(class_indices)
            n_test = int(n_class * test_size)
            test_idx.extend(class_indices[:n_test])
            train_idx.extend(class_indices[n_test:])
        
        train_idx = np.array(train_idx)
        test_idx = np.array(test_idx)
        np.random.shuffle(train_idx)
        np.random.shuffle(test_idx)
    
    X_train = X.iloc[train_idx].reset_index(drop=True)
    X_test = X.iloc[test_idx].reset_index(drop=True)
    y_train = y.iloc[train_idx].reset_index(drop=True)
    y_test = y.iloc[test_idx].reset_index(drop=True)

    return X_train, X_test, y_train, y_test, train_idx, test_idx

print(f"\n{'─' * 50}")
print("FUNCTIONS DEFINED")
print(f"{'─' * 50}")
print(" Scaler class")
print(" CategoricalEncoder class")
print(" split_data function")
````

````python
# =============================================================================
# EVALUATION METRICS
# =============================================================================

print("\n" + "=" * 70)
print("EVALUATION METRICS")
print("=" * 70)

# Calculate confusion matrix
def confusion_matrix(y_true, y_pred):

    y_true, y_pred = np.array(y_true), np.array(y_pred)
    tn = np.sum((y_true == 0) & (y_pred == 0))
    fp = np.sum((y_true == 0) & (y_pred == 1))
    fn = np.sum((y_true == 1) & (y_pred == 0))
    tp = np.sum((y_true == 1) & (y_pred == 1))
    return np.array([[tn, fp], [fn, tp]])

# Calculate accuracy
def accuracy(y_true, y_pred):
    return np.mean(np.array(y_true) == np.array(y_pred))

# Calculate precision: TP / (TP + FP)
def precision(y_true, y_pred):
    y_true, y_pred = np.array(y_true), np.array(y_pred)
    tp = np.sum((y_true == 1) & (y_pred == 1))
    fp = np.sum((y_true == 0) & (y_pred == 1))
    return tp / (tp + fp) if (tp + fp) > 0 else 0.0

# Calculate recall: TP / (TP + FN)
def recall(y_true, y_pred):
    y_true, y_pred = np.array(y_true), np.array(y_pred)
    tp = np.sum((y_true == 1) & (y_pred == 1))
    fn = np.sum((y_true == 1) & (y_pred == 0))
    return tp / (tp + fn) if (tp + fn) > 0 else 0.0

# Calculate F1 score
def f1_score(y_true, y_pred):
    p, r = precision(y_true, y_pred), recall(y_true, y_pred)
    return 2 * p * r / (p + r) if (p + r) > 0 else 0.0

# Calculate ROC-AUC using trapezoidal rule
def roc_auc(y_true, y_prob):
    y_true, y_prob = np.array(y_true), np.array(y_prob)
    sorted_idx = np.argsort(y_prob)[::-1]
    y_sorted = y_true[sorted_idx]
    
    n_pos, n_neg = np.sum(y_true == 1), np.sum(y_true == 0)
    if n_pos == 0 or n_neg == 0:
        return 0.5
    
    tpr, fpr = [0.0], [0.0]
    tp, fp = 0, 0
    
    for label in y_sorted:
        if label == 1:
            tp += 1
        else:
            fp += 1
        tpr.append(tp / n_pos)
        fpr.append(fp / n_neg)
    
    auc = sum((fpr[i] - fpr[i-1]) * (tpr[i] + tpr[i-1]) / 2 for i in range(1, len(fpr)))
    return auc

# Calculate ROC curve points
def roc_curve(y_true, y_prob):
    y_true, y_prob = np.array(y_true), np.array(y_prob)
    thresholds = np.sort(np.unique(y_prob))[::-1]
    n_pos, n_neg = np.sum(y_true == 1), np.sum(y_true == 0)
    
    fpr_list, tpr_list = [0.0], [0.0]
    for thresh in thresholds:
        y_pred = (y_prob >= thresh).astype(int)
        tp = np.sum((y_true == 1) & (y_pred == 1))
        fp = np.sum((y_true == 0) & (y_pred == 1))
        tpr_list.append(tp / n_pos if n_pos > 0 else 0)
        fpr_list.append(fp / n_neg if n_neg > 0 else 0)
    
    fpr_list.append(1.0)
    tpr_list.append(1.0)
    return np.array(fpr_list), np.array(tpr_list), thresholds

# Calculate precision-recall curve
def pr_curve(y_true, y_prob):
    y_true, y_prob = np.array(y_true), np.array(y_prob)
    thresholds = np.sort(np.unique(y_prob))[::-1]
    
    prec_list, rec_list = [1.0], [0.0]
    for thresh in thresholds:
        y_pred = (y_prob >= thresh).astype(int)
        tp = np.sum((y_true == 1) & (y_pred == 1))
        fp = np.sum((y_true == 0) & (y_pred == 1))
        fn = np.sum((y_true == 1) & (y_pred == 0))
        prec_list.append(tp / (tp + fp) if (tp + fp) > 0 else 1.0)
        rec_list.append(tp / (tp + fn) if (tp + fn) > 0 else 0.0)
    return np.array(prec_list), np.array(rec_list), thresholds

# Calculate average precision
def pr_auc(y_true, y_prob):
    prec, rec, _ = pr_curve(y_true, y_prob)
    return abs(sum((rec[i] - rec[i-1]) * prec[i] for i in range(1, len(rec))))

# Compute all metrics
def compute_all_metrics(y_true, y_pred, y_prob=None):
    metrics = {
        'Accuracy': accuracy(y_true, y_pred),
        'Precision': precision(y_true, y_pred),
        'Recall': recall(y_true, y_pred),
        'F1': f1_score(y_true, y_pred)
    }
    if y_prob is not None:
        metrics['ROC-AUC'] = roc_auc(y_true, y_prob)
        metrics['PR-AUC'] = pr_auc(y_true, y_prob)
    return metrics

print(f"\n{'─' * 50}")
print("FUNCTIONS DEFINED")
print(f"{'─' * 50}")
print(" confusion_matrix")
print(" accuracy, precision, recall, f1_score")
print(" roc_auc, roc_curve")
print(" pr_curve, pr_auc")
print(" compute_all_metrics")
````

## C(c) Working Prototype - Logistic Regression Implementation

**Selected Technique:** Logistic Regression

**Rationale:** Based on the Part B comparison, Logistic Regression is selected for detailed implementation due to:
- **Interpretability**: Coefficient-based feature importance for business stakeholders
- **Efficiency**: Fast training and inference suitable for production deployment
- **Probabilistic Output**: Confidence scores for risk stratification
- **Regulatory Compliance**: Transparent decision-making under GDPR/consumer protection laws
- **Class Imbalance Handling**: Built-in support for balanced class weighting

````python
# =============================================================================
# LOGISTIC REGRESSION MODEL  
# =============================================================================

print("\n" + "=" * 70)
print("LOGISTIC REGRESSION MODEL")
print("=" * 70)

class LogisticRegression:
    
    def __init__(self, learning_rate=0.01, max_iter=1000, class_weight='balanced', random_state=None):
        self.learning_rate = learning_rate
        self.max_iter = max_iter
        self.class_weight = class_weight
        self.random_state = random_state
        self.weights = None
        self.bias = 0.0
    
    def _sigmoid(self, z):
        return 1.0 / (1.0 + np.exp(-np.clip(z, -500, 500)))
    
    def _compute_class_weights(self, y):
        if self.class_weight != 'balanced':
            return np.ones(len(y))
        
        unique_classes, counts = np.unique(y, return_counts=True)
        n_samples = len(y)
        n_classes = len(unique_classes)
        
        class_weights = n_samples / (n_classes * counts)
        sample_weights = np.array([class_weights[int(label)] for label in y])
        return sample_weights
    
    def fit(self, X, y):
        if self.random_state is not None:
            np.random.seed(self.random_state)
        
        n_samples, n_features = X.shape
        self.weights = np.zeros(n_features)
        self.bias = 0.0
        
        sample_weights = self._compute_class_weights(y)
        
        for iteration in range(self.max_iter):
            linear_output = X @ self.weights + self.bias
            predictions = self._sigmoid(linear_output)
            
            errors = predictions - y
            weighted_errors = errors * sample_weights
            
            self.weights -= self.learning_rate * (X.T @ weighted_errors) / n_samples
            self.bias -= self.learning_rate * np.sum(weighted_errors) / n_samples
        
        return self
    # Return probabilities for both classes [P(class=0), P(class=1)]
    def predict_proba(self, X):
        linear_output = X @ self.weights + self.bias
        prob_positive = self._sigmoid(linear_output)
        prob_negative = 1 - prob_positive
        # Return 2D array with shape (n_samples, 2)
        return np.column_stack([prob_negative, prob_positive])
    
    def predict(self, X, threshold=0.5):
        probabilities = self.predict_proba(X)[:, 1]  # Get probability of positive class
        return (probabilities >= threshold).astype(int)

print(f"\n{'─' * 50}")
print("MODEL CLASS DEFINED")
print(f"{'─' * 50}")
print(" LogisticRegression class with:")
print("  Gradient descent optimization")
print("  Class weight balancing")
print("  Sigmoid activation")
print("  2D probability output")
````

````python
# ============================================================================
# PART E - FAIRNESS AND EDI ANALYSIS
# ============================================================================

print("\n" + "=" * 70)
print("PART E - EDI AND FAIRNESS ANALYSIS")
print("=" * 70)

# Get test data for demographic analysis (use X_test which contains the original features)
X_test_demo = X_test.copy()

fairness_metrics = {}

# Analyze by Gender (if available)
if 'gender' in X_test_demo.columns:
    for gender in X_test_demo['gender'].unique():
        mask = (X_test_demo['gender'] == gender).values
        y_true_group = y_test_arr[mask]
        y_pred_group = y_test_pred[mask]
        y_prob_group = y_test_prob[mask]
        metrics = compute_all_metrics(y_true_group, y_pred_group, y_prob_group)
        metrics['pos_rate'] = np.mean(y_pred_group)
        metrics['n'] = np.sum(mask)
        fairness_metrics[f'Gender_{gender}'] = metrics

# Analyze by Age Group (SeniorCitizen)
if 'SeniorCitizen' in X_test_demo.columns:
    for senior_status in [0, 1]:
        label = 'Senior' if senior_status == 1 else 'Non-Senior'
        mask = (X_test_demo['SeniorCitizen'] == senior_status).values
        y_true_group = y_test_arr[mask]
        y_pred_group = y_test_pred[mask]
        y_prob_group = y_test_prob[mask]
        if len(y_true_group) < 5:
            continue
        metrics = compute_all_metrics(y_true_group, y_pred_group, y_prob_group)
        metrics['pos_rate'] = np.mean(y_pred_group)
        metrics['n'] = np.sum(mask)
        fairness_metrics[f'Age_{label}'] = metrics

# Analyze by Partner status
if 'Partner' in X_test_demo.columns:
    for partner_status in X_test_demo['Partner'].unique():
        mask = (X_test_demo['Partner'] == partner_status).values
        y_true_group = y_test_arr[mask]
        y_pred_group = y_test_pred[mask]
        y_prob_group = y_test_prob[mask]
        if len(y_true_group) < 5:
            continue
        metrics = compute_all_metrics(y_true_group, y_pred_group, y_prob_group)
        metrics['pos_rate'] = np.mean(y_pred_group)
        metrics['n'] = np.sum(mask)
        fairness_metrics[f'Partner_{partner_status}'] = metrics

if fairness_metrics:
    # Visualize fairness metrics
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))
    
    groups = list(fairness_metrics.keys())
    
    # Determine colors by group type
    colors_acc = []
    for g in groups:
        if 'Gender' in g:
            colors_acc.append('#3498db')
        elif 'Age' in g:
            colors_acc.append('#e74c3c')
        else:
            colors_acc.append('#2ecc71')
    
    # Plot 1: Accuracy by Group
    ax1 = axes[0, 0]
    accuracies = [fairness_metrics[g]['Accuracy'] for g in groups]
    bars = ax1.bar(range(len(groups)), accuracies, color=colors_acc, edgecolor='black')
    ax1.set_xticks(range(len(groups)))
    ax1.set_xticklabels([g.split('_')[1] for g in groups], rotation=45, ha='right')
    ax1.set_ylabel('Accuracy')
    ax1.set_title('Accuracy by Demographic Group', fontweight='bold')
    ax1.axhline(test_metrics['Accuracy'], color='red', linestyle='--', 
                label=f'Overall: {test_metrics["Accuracy"]:.3f}')
    ax1.legend()
    ax1.set_ylim(0, 1.1)
    ax1.grid(axis='y', alpha=0.3)
    
    # Plot 2: Recall by Group
    ax2 = axes[0, 1]
    recalls = [fairness_metrics[g]['Recall'] for g in groups]
    bars = ax2.bar(range(len(groups)), recalls, color=colors_acc, edgecolor='black')
    ax2.set_xticks(range(len(groups)))
    ax2.set_xticklabels([g.split('_')[1] for g in groups], rotation=45, ha='right')
    ax2.set_ylabel('Recall')
    ax2.set_title('Recall by Demographic Group (Equal Opportunity)', fontweight='bold')
    ax2.axhline(test_metrics['Recall'], color='red', linestyle='--', 
                label=f'Overall: {test_metrics["Recall"]:.3f}')
    ax2.legend()
    ax2.set_ylim(0, 1.1)
    ax2.grid(axis='y', alpha=0.3)
    
    # Plot 3: Positive Prediction Rate
    ax3 = axes[1, 0]
    pos_rates = [fairness_metrics[g]['pos_rate'] for g in groups]
    bars = ax3.bar(range(len(groups)), pos_rates, color=colors_acc, edgecolor='black')
    ax3.set_xticks(range(len(groups)))
    ax3.set_xticklabels([g.split('_')[1] for g in groups], rotation=45, ha='right')
    ax3.set_ylabel('Positive Prediction Rate')
    ax3.set_title('Positive Prediction Rate (Demographic Parity)', fontweight='bold')
    overall_pos = np.mean(y_test_pred)
    ax3.axhline(overall_pos, color='red', linestyle='--', 
                label=f'Overall: {overall_pos:.3f}')
    ax3.legend()
    ax3.set_ylim(0, max(pos_rates) * 1.3)
    ax3.grid(axis='y', alpha=0.3)
    
    # Plot 4: Group Size
    ax4 = axes[1, 1]
    sizes = [fairness_metrics[g]['n'] for g in groups]
    bars = ax4.bar(range(len(groups)), sizes, color=colors_acc, edgecolor='black')
    ax4.set_xticks(range(len(groups)))
    ax4.set_xticklabels([g.split('_')[1] for g in groups], rotation=45, ha='right')
    ax4.set_ylabel('Sample Size')
    ax4.set_title('Sample Size by Demographic Group', fontweight='bold')
    ax4.grid(axis='y', alpha=0.3)
    
    plt.tight_layout()
    plt.savefig('fairness_analysis.png', dpi=150, bbox_inches='tight')
    plt.show()
    
    # Print fairness summary
    print("\nFairness Analysis Summary:")
    print("-" * 60)
    for group, metrics in fairness_metrics.items():
        print(f"\n{group} (n={metrics['n']}):")
        print(f"  Accuracy: {metrics['Accuracy']:.4f}")
        print(f"  Recall:   {metrics['Recall']:.4f}")
        print(f"  Precision: {metrics['Precision']:.4f}")
        print(f"  Pos Rate: {metrics['pos_rate']:.4f}")

print("\n Fairness analysis complete")
````
